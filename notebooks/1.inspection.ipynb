# Dataset inspection

A first look at the provided movie CSV.

```python
import json
import pandas as pd

data_path = "../data/original_dataset.csv"
# Load the dataset
movies = pd.read_csv(data_path)
movies.head()
```

## Dataset shape and columns

```python
# Check the dataset size and column names
print("Shape:", movies.shape)
print("Columns:")
print(movies.columns.tolist())
```

## Data types and missing values

```python
display(movies.dtypes.rename("dtype").to_frame())
# Check missing values
missing = movies.isna().sum().sort_values(ascending=False)
display(missing[missing > 0].rename("missing values").to_frame())
print("Total missing cells:", int(movies.isna().sum().sum()))
```

## Duplicates and unique values

```python
# Check duplicate rows and distinct values
print("Duplicate full rows:", int(movies.duplicated().sum()))
print("Rows with a repeated title:", int(movies["title"].duplicated(keep=False).sum()))
display(movies.nunique(dropna=False).rename("unique values").to_frame())
```

## Numerical columns

```python
# Summarize the numerical columns
numeric_columns = movies.select_dtypes(include="number").columns.tolist()
print("Numerical columns:", numeric_columns)
display(movies[numeric_columns].describe().T)
```

## Text and categorical fields

```python
# Look at a few examples from the text and category columns
text_columns = movies.select_dtypes(exclude="number").columns.tolist()
print("Text or categorical columns:", text_columns)
for column in ["title", "original_title", "original_language", "status"]:
    if column in movies.columns:
        print(f"\n{column} examples:", movies[column].dropna().astype(str).head(5).tolist())
```

## Inspecting structured metadata

```python
# Read a small example from the JSON metadata fields
def show_metadata_example(value):
    try:
        return json.loads(value)[:2]
    except (TypeError, json.JSONDecodeError):
        return []

for column in ["genres", "keywords", "production_companies", "production_countries", "spoken_languages"]:
    if column in movies.columns:
        print(f"{column}:", show_metadata_example(movies[column].dropna().iloc[0]))

display(movies[["title", "genres", "overview", "keywords", "release_date"]].head(5))
```

## Recommendation fields

```python
# Check which fields can support content recommendations
useful_columns = [column for column in ["title", "genres", "keywords", "overview", "tagline"]
                  if column in movies.columns]
print("Available content fields:", useful_columns)
print("Movie-level rating fields:", [column for column in ["vote_average", "vote_count"] if column in movies.columns])
print("User or user-movie interaction fields:", [column for column in movies.columns
      if "user" in column.lower() or "rater" in column.lower()])
print("Rows with genre metadata:", int(movies["genres"].fillna("").map(lambda x: bool(show_metadata_example(x))).sum()))
print("Rows with keyword metadata:", int(movies["keywords"].fillna("").map(lambda x: bool(show_metadata_example(x))).sum()))
```

## What I Found

- The CSV has 4,803 rows and 20 columns.
- The movie title field is `title`; genre labels and keyword labels are stored as JSON text in `genres` and `keywords`.
- `overview` and `tagline` provide additional text metadata. 4,775 rows have genre labels and 4,391 have keyword labels.
- There are 3,941 missing cells and 0 exact duplicate rows. No title is blank.
- The file has aggregate `vote_average` and `vote_count`, but no user-level ratings or interaction history.
- I will use genres, keywords, overview, and tagline for content-based recommendations.
